# **Neo4j official driver**

```python
from neo4j import GraphDatabase

uri = "bolt://127.0.0.1:7687"  
user = "neo4j"
password = "Default1234*"

driver = GraphDatabase.driver(uri, auth=(user, password))
```

```python
with driver.session() as session:
    print("Session opened")
```

```python
def create_company_graph(tx, company, year, metrics):
    # Create company node
    tx.run("MERGE (c:Company {name: $company})", company=company)
    
    # Create year node
    tx.run("MERGE (y:Year {value: $year})", year=year)
    
    # Connect company to year
    tx.run("""
        MATCH (c:Company {name:$company}), (y:Year {value:$year})
        MERGE (c)-[:REPORTED_IN]->(y)
    """, company=company, year=year)
    
    # Add financial metrics as nodes and connect
    for metric, value in metrics.items():
        if value is not None:
            tx.run("""
                MERGE (m:Metric {name:$metric, value:$value})
                WITH m
                MATCH (c:Company {name:$company})
                MERGE (c)-[:HAS_METRIC]->(m)
            """, metric=metric, value=float(value), company=company)
```

```python
with driver.session() as session:
    for idx, row in df.iterrows():
        company = row['Company ']
        year = int(row['Year'])
        metrics = {
            'Revenue': row['Revenue'],
            'Net Income': row['Net Income'],
            'EBITDA': row['EBITDA'],
            'Market Cap': row['Market Cap(in B USD)']
        }
        session.execute_write(create_company_graph, company, year, metrics)
```
